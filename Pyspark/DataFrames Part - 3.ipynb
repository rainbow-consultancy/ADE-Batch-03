```python
from pyspark.sql import functions as F
```

```python
# Read the Diamonds dataset

path = "/FileStore/tables/samples/diamonds.csv"

df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load(path)
df.display()
```

```python
# Pyspark Aggregations

new_df = (
    df
    .groupBy("cut")
    .count()
)
new_df.display()
```

```python
new_df = (
    df
    .groupBy("cut")
    .agg(
        F.count("*").alias("TotalRecords"),
        F.max(F.col("price")).alias("MaxPricedDiamond")
    )
)
new_df.display()
```

```python
# Read the dataframe

df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/FileStore/tables/Students_Social_Media_Addiction.csv")
)

display(df)
```

```python
new_df = (
    df.groupBy("Academic_Level")
    .agg(F.count("*").alias("NoOfStudents"))
)
new_df.display()
```

```python
# output - country, student_id_list 

# Banglasdesh [1,3,5,7,8]
# india [45, 6,7,8,9,100]

new_df = (
    df.groupBy("Country")
    .agg(
        F.collect_list("Student_ID").alias("student_id_list")
    )
)

new_df.display()
```

```python
# collect_set -- will not allow duplicate values in the list

new_df = (
    df.groupBy("Country")
    .agg(
        F.collect_set("Student_ID").alias("student_id_list")
    )
)

new_df.display()
```
