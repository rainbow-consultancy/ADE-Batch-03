```python
from pyspark.sql import functions as F
```

```python
# Read the dataframe

df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/FileStore/tables/Students_Social_Media_Addiction.csv")
)

display(df)
```

```python
# Adding/creating new columns in a dataframe

# withColumn() function is used to add a new column to a dataframe

new_df = (
    df
    .withColumn("is_active", F.lit(True))
)

new_df.display()
```

```python
new_df = (
    df
    .withColumn("is_active2", F.lit(True))
    .withColumn("is_active3", F.lit(False))
)

new_df.display()
```

```python
# Rename a column in a dataframe

# withColumnRenamed() function is used to rename a column in a dataframe

new_df = (
    df
    .withColumnRenamed("Student_ID", "stu_id")
)

new_df.display()
```

```python
new_df = (
    df
    .withColumnRenamed("Student_ID", "stu_id")
    .withColumnRenamed("Age", "age")
    .withColumnRenamed("Gender", "gender")
)

new_df.display()
```

```python
new_df = (
    df
    .withColumn(
        "age_category", 
        F.when(
            F.col("Age") < 20,
            "teenager"
        ).otherwise("adult")
    )
)

(
    new_df
    .select("Student_ID", "Age", "age_category")
    .display()
)
```

```python
new_df = (
    df
    .withColumn(
        "gender_2", 
        F.col("Gender")
    )
)

new_df.select("Student_ID", "Gender", "gender_2").display()
```

```python
# assign codes to relationship status 

# Single -- 100
# In Relationship -- 200
# Complicated -- 300


rel_df = (
    df
    .withColumn(
        "relationship_code",
        F.when(
            F.col("Relationship_Status") == "Single",
            100
        )
        .when(
            F.col("Relationship_Status") == "In Relationship",
            200
        )
        .when(
            F.col("Relationship_Status") == "Complicated",
            300
        )
        .otherwise(400)

    )
)

rel_df.select("Student_ID", "Relationship_Status", "relationship_code").display()
```

```python
# dropDuplicates  -- by default is behaves similar to distinct operation

# default behaviour

new_df = df.dropDuplicates()
new_df.count()
```

```python
# using subset of columns

new_df = df.dropDuplicates(subset=["Relationship_Status"])
new_df.count()
```

```python
new_df.display()
```

```python
# drop function will delete a column from the dataframe

new_df = df.drop("Addicted_Score")
new_df.printSchema()
```

```python
# Handling Null values in dataframe
# since this df doesn't have nulls I'll create a new column and insert nulls in it

new_df = (
    df
    .withColumn(
        "is_active",
        F.when(
            F.col("Student_ID") % 2 != 0,
            F.lit(None)
        ).otherwise(1)
    )
)

new_df.select("Student_ID", "Age", "is_active").display()
```

```python
# let's say your business came and ask you like if column is_active is null then drop those rows
# by default if you don't pass any subset of columns it will delete the rows where the columns have null values in it

null_df = new_df.dropna()
```

```python
null_df = new_df.dropna(subset=["is_active"])
null_df.count()
```

```python
null_df.select("Student_ID", "Age", "is_active").display()
```

```python
# let's say your business came and ask you like if column is_active is null then make that null value to 100

null_df = new_df.fillna(100, subset=["is_active"])
null_df.select("Student_ID", "Age", "is_active").display()
```

```python
null_df = new_df.na.fill(100, subset=["is_active"])
null_df.select("Student_ID", "Age", "is_active").display()
```
